```python
import matplotlib.pyplot as plt
from scipy import stats

x = [1,2,3,4,5]
y = [10,17,37,32,50]

slop,intercept,r,p,std_err = stats.linregress
def myfunc(x):
    return slope * x + intercept
mymodel = list(map(myfunc, x))
    
plt.scatter(x,y)
plt.plot(x,mymodel)
plt.show()
```
